## Verify the figures outside the dashboard
This creates the same graphs directly using Plotly, useful if you need to validate the dataset before launching the Dash application.

````python
import plotly.express as px
success = data.query("`class` == 1").groupby("Launch Site").size().reset_index(name="Successful Landings")
px.pie(success, values="Successful Landings", names="Launch Site", title="All-site success count").show()
````

````python
site = data["Launch Site"].unique()[0]
part = data[data["Launch Site"] == site]
counts = part["class"].value_counts().reindex([0,1],fill_value=0)
px.pie(names=["Failure","Success"],values=counts.values,title=f"Outcomes: {site}").show()
````

````python
px.scatter(data[data["Payload Mass (kg)"].between(0,10000)],x="Payload Mass (kg)",y="class",color="Booster Version Category",title="Payload versus landing outcome").show()
````
